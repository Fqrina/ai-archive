```python
col = [
#     'id', 'expedition_name', 
'exploration_number', 'total_laps',
       'total_distance', 
       'exploration_zone', 
       'terrain_type',
       'route_orientation', 'lap_distance', 'navigation_points',
       'planet_radius', 'field_coord_x', 'field_coord_y', 'field_coord_z',
       'exploration_planet', 
       'exploration_planet_sector', 
       'explorer_name',
       'explorer_biometric_age',
       'birth_planet', 
       'birth_planet_sector',
       'citizenship_planet', 
       'citizenship_sector', 
       'explorer_team',
       'explorer_team_planet', 
       'explorer_team_sector', 'engine_team',
       'engine_team_planet', 'engine_team_cluster', 'simulation_time_alpha',
       'simulation_laps_alpha', 'simulation_time_beta', 'simulation_laps_beta',
       'simulation_time_gamma', 'simulation_laps_gamma', 'env_test_time_a',
       'env_test_time_b', 'env_test_time_c', 'total_test_laps', 'launch_delay',
       'launch_delay_offset', 'initial_launch_order', 'final_launch_order',
       'field_demo_laps', 'field_demo_time', 'field_calib_time_a',
       'field_calib_time_b', 'field_calib_time_c', 'field_calib_laps',
    #    'mission_duration', 'pit_stops', 'mission_speed_rank', 
       'mission_award']
def tostr(x):
    if pd.isna(x):
        return -1
    else:
        if x == "SFB":
            return 0
        return int(x)
train["launch_delay"] = train["launch_delay"].apply(tostr)
```

```python
def applydate(X):
    if pd.isna(X) or X == -1 or str(X).strip() == "-1":
        return -1
    try:
        val = str(X).strip()
        d = 0
        if " " in val:
            d_part, val = val.split(" ", 1)
            d = d_part.replace("d", "")
        parts = val.split(":")
        if len(parts) == 3:
            h, m, s_ms = parts[0], parts[1], parts[2]
        elif len(parts) == 2:
            h, m, s_ms = 0, parts[0], parts[1]
        elif len(parts) == 1:
            h, m, s_ms = 0, 0, parts[0]
        else:
            return -1
        if "." in str(s_ms):
            s, ms = str(s_ms).split(".")
        else:
            s, ms = s_ms, 0
        total_seconds = (
            86400 * int(d) + 
            3600 * int(h) + 
            60 * int(m) + 
            int(s) + 
            float(int(ms) / 1000)
        )
        return total_seconds
    except Exception as e:
        print("error", X, e)
        return -1

train["simulation_time_alpha"] = train["simulation_time_alpha"].apply(applydate)
train["simulation_time_beta"] = train["simulation_time_beta"].apply(applydate)
train["simulation_time_gamma"] = train["simulation_time_gamma"].apply(applydate)
train["env_test_time_a"] = train["env_test_time_a"].apply(applydate)
train["env_test_time_b"] = train["env_test_time_b"].apply(applydate)
train["field_demo_time"] = train["field_demo_time"].apply(applydate)
train["env_test_time_c"] = train["env_test_time_c"].apply(applydate)
# train["mission_duration"] = train["mission_duration"].apply(applydate)
```
